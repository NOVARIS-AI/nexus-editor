```python
# Small, bounded GPU smoke test (does not download a model)
import torch, time
if torch.cuda.is_available():
    device = 'cuda'
    x = torch.randn((2048, 2048), device=device)
    torch.cuda.synchronize()
    start = time.time()
    y = x @ x
    torch.cuda.synchronize()
    print('GPU smoke test passed; seconds:', round(time.time() - start, 3))
    print('Result finite:', bool(torch.isfinite(y).all().item()))
    del x, y
    torch.cuda.empty_cache()
else:
    print('No GPU assigned. In Colab choose Runtime > Change runtime type > T4 GPU if available. In Kaggle choose Settings > Accelerator > GPU.')
```
